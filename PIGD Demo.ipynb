```python
import plotly.express as px
from ipywidgets import HTML, VBox, Image
from plotly import graph_objects as go
#from IPython.display import Image

"""Following code based on example here: https://gist.github.com/raphaeljolivet/f076256e589f67c028a6bffaab279601"""

def interactive_plot(df, fig, template, event="hover") :
    """
    Make a plot react on hover or click of a data point and update a HTML preview below it.
    **template** Should be a string and contain placeholders like {colname} to be replaced by the value
    of the corresponding data row.

    """

    file = open("/content/drive/My Drive/placeholder.jpg","rb") # WHAT DATE
    image = file.read()
    img = Image(
        value=image,
        format='jpg',
        width=600,
        height=800
    )

    def update(trace, points, state):
        ind = points.point_inds[0]
        date = df.loc[ind,'Date']
        file = open(template.format(id=date), "rb")
        image = file.read()
        img.value = image

    fig2 = go.FigureWidget(data=fig.data, layout=fig.layout)

    if event == "hover" :
        fig2.data[0].on_hover(update)
    else :
        fig2.data[0].on_click(update)

    #print(template)

    return VBox([fig2, img])
```

```python
plot=px.line(df,'Hours From Start', 'Area')
template='/content/drive/My Drive/4091_images/{id}/base_{id}.jpg'
interactive_plot(df, plot, template, event="click")
```

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import numpy as np
from matplotlib import rc
from PIL import Image

#fig = plt.figure(figsize=(10,10))
fig, (ax1, ax2) = plt.subplots(2, 1)
#ax1.set_xlim(0,11000)
#ax1.set_ylim(0,2000000)
ax1.set_xlabel('Hours From Start', fontsize=15)
ax1.set_ylabel('Area (pixels)', fontsize=15)
ax1.set_title('Plant Growth')

tmp = df.iloc[:1,:]
scat = ax1.scatter(tmp['Hours From Start'], tmp['Area'])

template='/content/drive/My Drive/4091_images/{id}/base_{id}.jpg'
image = np.asarray(Image.open(template.format(id=df.loc[0,'Date'])))
img = ax2.imshow(image)

def animate(i):
    tmp = df.iloc[:i,:]
    data = np.stack([tmp['Hours From Start'],tmp['Area']]).T
    #scat.set_offsets(tmp['Hours From Start'], tmp['Area'], s=tmp['Group'], c=tmp['Group'], ec='k')
    scat.set_offsets(data)
    image = np.asarray(Image.open(template.format(id=df.loc[0,'Date'])))
    img = ax2.imshow(image)
    return (scat, img)

anim = FuncAnimation(fig, animate, frames=len(df.index), interval=30, repeat=False)
#anim.save('gdp_life.gif', writer='Pillow')
#plt.show()

rc('animation', html='jshtml')
anim
```
